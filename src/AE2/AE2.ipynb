## 1. Importación de librerias y de las clases que se piden en el enunciado

```python
from pathlib import Path
from manejador_csv import ManejadorCSV
from manejador_json import ManejadorJSON
from generador_resumen_txt import GeneradorResumenTXT

print("Las tres clases se importaron correctamente.")
```

## 2. Defición de las rutas donde se encuentrar los archivos csv y json

```python
BASE = Path.cwd()
DATOS = BASE / "datos"

CSV_ORIGINAL = DATOS / "mejores_barrios_precio_calidad_bogota_datos.csv"
JSON_ORIGINAL = DATOS / "Adult_income.json"

CSV_MODIFICADO = DATOS / "barrios_modificado.csv"
JSON_MODIFICADO = DATOS / "adult_income_modificado.json"
TXT_RESUMEN = DATOS / "resumen.txt"

URL_CSV = "https://huggingface.co/datasets/aptologia/mejores-barrios-precio-calidad-bogota-2026"
URL_JSON = "https://huggingface.co/datasets/diabolic6045/Adult-Income-Json"

print("Carpeta de trabajo:", BASE)
print("CSV original encontrado:", CSV_ORIGINAL.exists())
print("JSON original encontrado:", JSON_ORIGINAL.exists())
```

## 3. Se crean los objetos y se revisan los datos dentro de los archivos

```python
# Crear los objetos de las clases
manejador_csv = ManejadorCSV()
manejador_json = ManejadorJSON()

# Importar los archivos originales
encabezados_csv, datos_csv = manejador_csv.importar(CSV_ORIGINAL)
datos_json = manejador_json.importar(JSON_ORIGINAL)

# Mostrar un resumen de la importación
print("CSV importado correctamente")
print("Columnas:", encabezados_csv)
print("Cantidad de registros CSV:", len(datos_csv))

print("\nJSON importado correctamente")
print("Cantidad de registros JSON:", len(datos_json))

# Mostrar solo los primeros dos registros para comprobar su estructura
print("\nPrimer registro CSV:")
print(datos_csv[0] if datos_csv else "No hay registros CSV")

print("\nPrimer registro JSON:")
print(datos_json[0])
```

## 4. Se realizan las modificaciones a los archivos y se guardan en el nuevo destino

```python
# 4.1. Mostrar los datos antes de modificarlos
print("=== ANTES DE LAS MODIFICACIONES ===")
print("CSV - Cantidad de columnas:", len(manejador_csv.encabezados))
print("CSV - Primer registro:", manejador_csv.datos[0])
print("JSON - Primer registro:", manejador_json.datos[0])

# 4.2. Modificar cada archivo mediante su propia clase
manejador_csv.modificar()
manejador_json.modificar()

# 4.3. Guardar las nuevas versiones
ruta_csv_guardado = manejador_csv.guardar(CSV_MODIFICADO)
ruta_json_guardado = manejador_json.guardar(JSON_MODIFICADO)

# 4.4. Mostrar los resultados
print("\n=== DESPUÉS DE LAS MODIFICACIONES ===")
print("CSV - Cantidad de columnas:", len(manejador_csv.encabezados))
print("CSV - Primer registro:", manejador_csv.datos[0])
print("Modificación CSV:", manejador_csv.modificacion)

print("\nJSON - Primer registro:", manejador_json.datos[0])
print("Modificación JSON:", manejador_json.modificacion)

# 4.5. Se guardan los nuevos archivos y se muestra una comprobación de que se guardaron correctamente
print("\n=== ARCHIVOS GUARDADOS ===")
print("CSV:", ruta_csv_guardado)
print("JSON:", ruta_json_guardado)
print("CSV guardado correctamente:", CSV_MODIFICADO.exists())
print("JSON guardado correctamente:", JSON_MODIFICADO.exists())
```

## 5. Se genera el resumen TXT

```python
generador = GeneradorResumenTXT()

contenido_resumen = generador.generar(
    url_csv=URL_CSV,
    url_json=URL_JSON,
    archivo_csv_original=str(CSV_ORIGINAL),
    archivo_json_original=str(JSON_ORIGINAL),
    archivo_csv_modificado=str(CSV_MODIFICADO),
    archivo_json_modificado=str(JSON_MODIFICADO),
    registros_csv=len(datos_csv),
    registros_json=len(datos_json),
    modificacion_csv=manejador_csv.modificacion,
    modificacion_json=manejador_json.modificacion,
)

ruta_resumen = generador.guardar(DATOS / "resumen.txt")

print("Resumen generado en:", ruta_resumen)
print("\n=== CONTENIDO DEL RESUMEN ===\n")
print(contenido_resumen)
```

## 6. Validación final

```python
print("=== VALIDACIÓN FINAL ===")

for archivo in [CSV_MODIFICADO, JSON_MODIFICADO, TXT_RESUMEN]:
    if archivo.exists():
        print(f"OK: {archivo.name}")
    else:
        print(f"ERROR: No se encontró {archivo.name}")
```
