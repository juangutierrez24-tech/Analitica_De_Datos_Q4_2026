## 1. Preparar el entorno e importar la clase de carga

El notebook puede ejecutarse desde distintas carpetas. Por eso se buscan los directorios padres hasta encontrar `src/clases_po` y se agrega `src` a `sys.path`. Así se puede importar `HFDatasetLoader` sin instalar el proyecto como paquete.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
while ROOT != ROOT.parent and not (ROOT / "src" / "clases_po").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT / "src"))

from clases_po import HFDatasetLoader
```

## 2. Configurar el dataset y la salida

`dataset_name` identifica el repositorio en Hugging Face y `split="train"` indica la división que se descargará. El parámetro `filename` hace que el `DataFrame` también se guarde como CSV en `src/sincronicas/datos/` cuando el notebook se ejecuta desde esa carpeta.

```python
loader = HFDatasetLoader(
    dataset_name="aptologia/precio-m2-venta-apartamentos-ciudades-latinoamerica-2026",
    split="train",
    filename="datos/precio_m2_apartamentos.csv",
)
```

## 3. Descargar el dataset y crear el DataFrame

El método `load()` descarga el dataset, lo convierte a `pandas.DataFrame` y lo asigna a la variable `df`. La expresión `df.head()` muestra las primeras filas para comprobar que la carga fue exitosa.

```python
df = loader.load()
df.head()
```

## 4. Revisar la estructura del DataFrame

Esta celda realiza una revisión inicial: cuenta filas y columnas y muestra el tipo de cada columna junto con información de memoria y valores no nulos. Esta inspección ayuda a detectar problemas antes de analizar los datos.

```python
print(f"Filas: {len(df)}")
print(f"Columnas: {len(df.columns)}")
df.info()
```

## 5. Transformaciones del dataset

```python
import pandas as pd
from IPython.display import display

# Trabajamos sobre una copia para conservar los datos originales
df_transformado = df.copy()

# Mostrar las columnas disponibles
print("Columnas del dataset:")
print(df_transformado.columns.tolist())

print("\nDimensiones originales:", df_transformado.shape)
display(df_transformado.head())
```

## 5.1 Transformación 1: limpieza de datos.

```python
# Eliminar filas completamente duplicadas
duplicados = df_transformado.duplicated().sum()
df_transformado = df_transformado.drop_duplicates()

# Identificar valores faltantes
faltantes = df_transformado.isnull().sum()

print("Filas duplicadas eliminadas:", duplicados)
print("\nValores faltantes por columna:")
display(faltantes.to_frame(name="Cantidad de faltantes"))

print("\nDimensiones después de la limpieza:", df_transformado.shape)
```

## 5.2 Transformación 2: clasificación de precios.

```python
print("Columnas que podrían contener el precio por metro cuadrado:")

for columna in df_transformado.columns:
    nombre = str(columna).lower()

    if "precio" in nombre or "price" in nombre or "m2" in nombre:
        print("-", columna)
```

## 5.3 Transformación 3: agrupación y comparación por país.

```python
print("Columnas que podrían identificar el país:")

for columna in df_transformado.columns:
    nombre = str(columna).lower()

    if "pais" in nombre or "país" in nombre or "country" in nombre:
        print("-", columna)
```
